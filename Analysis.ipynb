```python
#import libraries
import pandas as pd;
import numpy as np;

#visualizing data

import matplotlib.pyplot as plt 
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv('Sales Data.csv', encoding = 'unicode_escape')
```

```python
df.shape
```

```python
df.head(10)
```

```python
df.info()
```

```python
#drop blanks \ unrelated columns

df.drop(['Status','unnamed1'],axis = 1, inplace = True)
```

```python
df.info()
```

```python
pd.isnull(df)
```

```python
#check for null vales
pd.isnull(df).sum()
```

```python
#drop null values
df.dropna(inplace=True)
```

```python
df.shape
```

```python
#check the null values again it show 0 now
pd.isnull(df).sum()
```

```python
#change data type
df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
df.rename(columns = {'Cust_name':'Customer_name'}, inplace = True)
```

```python
#describe() method returns description of the data in the dataframe
df[['Age','Orders','Amount']].describe()
```

**Exploratory Data Analysis**

**Gender**

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender', data = df, hue = 'Gender',palette=['Blue', 'pink'],
    legend=False) 
#it shows values in chart
#sns.countplot() create a bar chart shows number of people in each gender category
#ax.containers - gets the bars
for bars in ax.containers:
    ax.bar_label(bars)
#ax.bar-label(bars)- count value on top of each bar
```

```python
df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending = False)
```

```python
high_sales = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending = False)

sns.barplot(x = 'Gender', y = 'Amount', data = high_sales, hue = 'Gender')
```

**from above graphs we can see that most of the buyers are females**

**Age**

```python
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

**From above graphs we can see the most of the buyers are the age group between 26-35 years female**

**STATE**

```python
state_count = df['State'].value_counts()

sns.barplot(x = state_count.values, y = state_count.index)
```

**from above chart we can see how many customers are from each state**

**Occupation**

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Occupation', hue = 'Occupation', legend = False)

for bars in ax.containers:
    ax.bar_label(bars)
```

**From above graphs we can see the most of the buyers are working in IT, Aviation and HealthCare sector**

**Product Category**

```python
plt.figure(figsize=(10,8))
ax = sns.countplot(data = df, y = 'Product_Category', hue = 'Product_Category', legend = False)
for bars in ax.containers:
    ax.bar_label(bars)
plt.tight_layout()
```

**Amount**

```python
sns.histplot(df['Amount'], bins = 20,kde = True)
```

**This Histogram shows the distribution of purchase amounts to understand where the most of the purchase amount are concentrated**

**Zone**

```python
df['Zone'].value_counts().plot(kind = 'pie', autopct = '%1.2f%%',labeldistance=1.1)
plt.ylabel('')# remove the unwanted name ex count
plt.tight_layout()# moves lit outsider pie
```

**Conclusion & Key Insights**

*Identified customer purchasing patterns by gender, state ,zone.  
*Analyzed the most popular product category.  
*The Zone analysis shows the distribution of customers across different regions.  
*These insights can help businesses understand customer pattern and make data driven decisions.
